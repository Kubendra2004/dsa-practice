# LeetCode 560: Subarray Sum Equals K

## Problem Statement
Given an integer array `nums` and an integer `k`, return the number of continuous subarrays whose sum equals `k`.

## Concept
**Prefix Sum + HashMap** — count previous prefix sums and look up `current_sum - k`.

## Difficulty
Medium

## Constraints
- 1 ≤ len(nums) ≤ 2 × 10^4
- -1000 ≤ nums[i] ≤ 1000
- -10^7 ≤ k ≤ 10^7

## Examples
```
Input: nums = [1,1,1], k = 2
Output: 2
Explanation: There are two subarrays that sum to 2.
```

````python
def subarray_sum_equals_k(nums, k):
    """
    Counts the number of continuous subarrays whose sum equals k.
    
    Args:
        nums: Input list of integers
        k: Target sum
    
    Returns:
        Number of subarrays with sum equal to k
    """
    count=0
    cursum=0
    presum={}
    presum[0]=1
    for i in range(len(nums)):
        cursum+=nums[i]
        target=cursum-k
        if target in presum:
            count+=presum[target]
        if cursum in presum:
            pastfreq=presum[cursum]
            presum[cursum]=pastfreq+1
        else:
            presum[cursum]=1
    return count
````

````python
# Test Cases

nums1 = [1, 1, 1]
k1 = 2
print(f"Test 1 - Input: {nums1}, k={k1}")
print("Expected: 2")
print(f"Got: {subarray_sum_equals_k(nums1, k1)}")
print()

nums2 = [1, 2, 3]
k2 = 3
print(f"Test 2 - Input: {nums2}, k={k2}")
print("Expected: 2")
print(f"Got: {subarray_sum_equals_k(nums2, k2)}")
print()

nums3 = [1, -1, 0]
k3 = 0
print(f"Test 3 - Input: {nums3}, k={k3}")
print("Expected: 3")
print(f"Got: {subarray_sum_equals_k(nums3, k3)}")
````
