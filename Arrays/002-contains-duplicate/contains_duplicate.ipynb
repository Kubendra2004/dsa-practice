# LeetCode 217: Contains Duplicate

## Problem Statement
Given an integer array `nums`, return `True` if any value appears **at least twice** in the array, and return `False` if every element is distinct.

## Concept
**Hashing / Set-based deduplication** — efficient duplicate detection

## Difficulty
Easy

## Constraints
- 1 ≤ len(nums) ≤ 10⁵
- -10⁹ ≤ nums[i] ≤ 10⁹

## Examples
```
Input: nums = [1, 2, 3, 1]
Output: True
Explanation: The value 1 appears twice.

Input: nums = [1, 2, 3, 4]
Output: False
Explanation: All elements are distinct.
```

````python
def contains_duplicate(nums):
    """
    Determines if the array contains any duplicate values.
    
    Args:
        nums: List of integers
    
    Returns:
        True if duplicates exist, False otherwise
    """
    return len(nums) != len(set(nums))
````

````python
# Test Cases

# Test 1: Array with duplicates
nums1 = [1, 2, 3, 1]
print(f"Test 1 - Input: {nums1}")
print(f"Expected: True")
print(f"Got: {contains_duplicate(nums1)}")
print()

# Test 2: Array with all distinct elements
nums2 = [1, 2, 3, 4]
print(f"Test 2 - Input: {nums2}")
print(f"Expected: False")
print(f"Got: {contains_duplicate(nums2)}")
print()

# Test 3: Single element
nums3 = [1]
print(f"Test 3 - Input: {nums3}")
print(f"Expected: False")
print(f"Got: {contains_duplicate(nums3)}")
print()

# Test 4: Large duplicates
nums4 = [99, 99]
print(f"Test 4 - Input: {nums4}")
print(f"Expected: True")
print(f"Got: {contains_duplicate(nums4)}")
````
