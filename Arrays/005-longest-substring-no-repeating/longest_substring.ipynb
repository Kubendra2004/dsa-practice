# LeetCode 3: Longest Substring Without Repeating Characters

## Problem Statement
Given a string `s`, find the length of the **longest substring** without repeating characters.

## Concept
**Sliding Window with HashMap** — expand the window as you scan, and move the left boundary past the last seen duplicate when needed.

## Difficulty
Easy

## Constraints
- 0 ≤ len(s) ≤ 5 × 10⁴
- s consists of English letters, digits, symbols, and spaces

## Examples
```
Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc" (length 3).

Input: s = "bbbbb"
Output: 1
Explanation: The answer is "b" (length 1).

Input: s = "pwwkew"
Output: 3
Explanation: The answer is "wke" (length 3).
```

````python
def length_of_longest_substring(s):
    """
    Finds the length of the longest substring without repeating characters.
    
    Args:
        s: Input string
    
    Returns:
        Length of the longest substring without repeating characters
    """
    if s is None:
        return 0

    last_seen = {}
    left = 0
    max_len = 0

    for right, ch in enumerate(s):
        if ch in last_seen:
            left = max(left, last_seen[ch] + 1)
        last_seen[ch] = right
        max_len = max(max_len, right - left + 1)

    return max_len
````

````python
# Test Cases

# Test 1: Multiple repeating characters
s1 = "abcabcbb"
print(f"Test 1 - Input: \"{s1}\"")
print(f"Expected: 3 (substring \"abc\")")
print(f"Got: {length_of_longest_substring(s1)}")
print()

# Test 2: All same characters
s2 = "bbbbb"
print(f"Test 2 - Input: \"{s2}\"")
print(f"Expected: 1 (substring \"b\")")
print(f"Got: {length_of_longest_substring(s2)}")
print()

# Test 3: Mixed repeating pattern
s3 = "pwwkew"
print(f"Test 3 - Input: \"{s3}\"")
print(f"Expected: 3 (substring \"wke\")")
print(f"Got: {length_of_longest_substring(s3)}")
print()

# Test 4: Empty string
s4 = ""
print(f"Test 4 - Input: \"{s4}\"")
print(f"Expected: 0 (empty string)")
print(f"Got: {length_of_longest_substring(s4)}")
print()

# Test 5: All unique characters
s5 = "abcdefg"
print(f"Test 5 - Input: \"{s5}\"")
print(f"Expected: 7 (all characters)")
print(f"Got: {length_of_longest_substring(s5)}")
````
