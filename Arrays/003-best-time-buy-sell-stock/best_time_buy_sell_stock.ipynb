# LeetCode 121: Best Time to Buy and Sell Stock

## Problem Statement
You are given an array `prices` where `prices[i]` is the price of a given stock on the i-th day. You want to **maximize profit** by choosing a single day to buy and a **different later day** to sell. Return the **maximum profit** you can achieve. If you can't make any profit, return `0`.

## Concept
**One-pass tracking / Greedy observation** — efficient solution in linear time

## Difficulty
Easy

## Constraints
- 1 ≤ len(prices) ≤ 10⁵
- 0 ≤ prices[i] ≤ 10⁴

## Examples
```
Input: prices = [7, 1, 5, 3, 6, 4]
Output: 5
Explanation: Buy at 1, sell at 6 → profit = 6 - 1 = 5.

Input: prices = [7, 6, 4, 3, 1]
Output: 0
Explanation: No profit possible; prices only decrease.
```

````python
def maxprofit(prices):
    """
    Finds the maximum profit from buying and selling a stock once.
    
    Args:
        prices: List of integers representing stock prices per day
    
    Returns:
        The maximum profit possible, or 0 if no profit can be made
    """
    # TODO: Implement your solution here

    min_price = float('inf')
    max_profit = 0
        
    for price in prices:
        
        if price < min_price:
            min_price = price
        else:
            profit = price - min_price
            if profit > max_profit:
                max_profit = profit
                    
    return max_profit
````

````python
# Test Cases

# Test 1: Mixed prices with profit opportunity
prices1 = [7, 1, 5, 3, 6, 4]
print(f"Test 1 - Input: {prices1}")
print(f"Expected: 5 (buy at 1, sell at 6)")
print(f"Got: {maxprofit(prices1)}")
print()

# Test 2: Descending prices (no profit possible)
prices2 = [7, 6, 4, 3, 1]
print(f"Test 2 - Input: {prices2}")
print(f"Expected: 0 (prices only decrease)")
print(f"Got: {maxprofit(prices2)}")
print()

# Test 3: Only two prices
prices3 = [2, 4]
print(f"Test 3 - Input: {prices3}")
print(f"Expected: 2 (buy at 2, sell at 4)")
print(f"Got: {maxprofit(prices3)}")
print()

# Test 4: Best buy/sell at edges
prices4 = [1, 7, 5, 11, 5]
print(f"Test 4 - Input: {prices4}")
print(f"Expected: 10 (buy at 1, sell at 11)")
print(f"Got: {maxprofit(prices4)}")
````
